```python
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
N = 10_000  # total points per dataset


def make_linear_dataset(n=N, seed=SEED):
    """Two well-separated Gaussian clusters (linearly separable)."""
        rng = np.random.default_rng(seed)
            half = n // 2
                X0 = rng.normal(loc=[-4, -4], scale=1.0, size=(half, 2))
                    X1 = rng.normal(loc=[4, 4], scale=1.0, size=(half, 2))
                        X = np.vstack([X0, X1])
                            y = np.array([0] * half + [1] * half)
                                return X, y


                                def make_circular_dataset(n=N, seed=SEED):
                                    """Class 0: Gaussian blob in the center. Class 1: ring around it."""
                                        rng = np.random.default_rng(seed)
                                            half = n // 2
                                                # inner cluster
                                                    X0 = rng.normal(loc=[0, 0], scale=0.8, size=(half, 2))
                                                        # outer cluster: random angle, radius around 4
                                                            angle = rng.uniform(0, 2 * np.pi, size=half)
                                                                radius = rng.normal(loc=4.0, scale=0.7, size=half)
                                                                    X1 = np.column_stack([radius * np.cos(angle), radius * np.sin(angle)])
                                                                        X = np.vstack([X0, X1])
                                                                            y = np.array([0] * half + [1] * half)
                                                                                return X, y


                                                                                def plot_dataset(ax, X, y, title):
                                                                                    ax.scatter(X[y == 0, 0], X[y == 0, 1], s=5, alpha=0.5, label="Class 0")
                                                                                        ax.scatter(X[y == 1, 0], X[y == 1, 1], s=5, alpha=0.5, label="Class 1")
                                                                                            ax.set_xlabel("X1")
                                                                                                ax.set_ylabel("X2")
                                                                                                    ax.set_title(title)
                                                                                                        ax.set_aspect("equal")
                                                                                                            ax.legend(markerscale=3)


                                                                                                            X_lin, y_lin = make_linear_dataset()
                                                                                                            X_circ, y_circ = make_circular_dataset()

                                                                                                            fig, axes = plt.subplots(1, 2, figsize=(11, 5))
                                                                                                            plot_dataset(axes[0], X_lin, y_lin, "(a) Linear dataset")
                                                                                                            plot_dataset(axes[1], X_circ, y_circ, "(b) Circular dataset")
                                                                                                            plt.tight_layout()
                                                                                                            plt.show()
```
